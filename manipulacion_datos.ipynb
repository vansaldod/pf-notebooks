## Introducción al DataFrame

Un Dataframe es un objeto tabulas, organizado por filas y columna, un
DataFrame de dimensión **(n, m)** tiene **n filas** y **m columnas**.
Cada columna puede tener un tipo de dato diferente.

Sus elementos son:

- **Índice:** identifica cada fila.
- **Columnas:** identifican cada variable.
- **Valores:** datos en la intersección de una fila y una columna.



## Introducción a Pandas

Pandas es una librería de Python para **manipulación y análisis de
datos**. Muchas veces se define a Pandas como un **Excel en Python**, ya
que es la plataforma más utilizada para trabajar con datos tabulares.

La convención para importar Pandas es:


```python
import pandas as pd
```

## Formas de crear un DataFrame

Existen varias formas de crear un DataFrame. En este curso nos vamos a
enfocar en tres:

1)  Crear un DataFrame a partir de un **diccionario** de listas.
2)  Cargar un DataFrame desde un **archivo Excel o CSV**.
3)  Descargar un DataFrame desde una **API**.

## Crear un DataFrame desde un diccionario de listas

- **Lista:** secuencia ordenada de elementos, escrita entre corchetes:
  `[34, 45]`.
- **Diccionario:** asocia claves con valores, entre llaves:
  `{"ACCION_A": [34, 45]}`.


```python
import pandas as pd
data_precios = {
  "ACCION_A": [34, 45],
  "ACCION_B": [12, 23]
}
df = pd.DataFrame(data_precios)
print(df)
```

```python
import pandas as pd
data_precios = {
  "ACCION_A": [34, 45],
  "ACCION_B": [12, 23]
}
df = pd.DataFrame(data_precios)
print(df)
```

## Cargar CSV

Los archivos CSV (Comma Separated Values) son archivos de texto plano
que contienen datos tabulares. Cada línea del archivo representa una
fila de la tabla y los valores de cada columna están separados por
comas. Este formato es ampliamente utilizado para la generación y
lectura de datos, ya que es compatible con la mayoría de los programas
de hojas de cálculo y bases de datos, además de que no tiene limites de
tamaño y es un formato más liviano que un archivo de Excel, aunque para
grandes hay otros formatos más eficientes.

En este ejemplo, el archivo `precios_portafolio.csv` contiene los mismos
datos que el archivo de Excel anterior, pero en formato CSV.


```python
import pandas as pd
file_csv = 'precios_portafolio.csv'
df = pd.read_csv(file_csv, sep=',')
print(df)
```

## API y SDK

Para poder acceder a distintos tipos de información existen proveedores
que permiten el acceso a sus bases de datos mediante una **API**
(Application Programming Interface). Una API es un conjunto de reglas y
protocolos que permiten que diferentes aplicaciones se comuniquen entre
sí.

Para poder consular una api desde un script de python hay que
dessarrollar una infraestructura de código que haga las consultas y
reciba los datos. Para simplificar este proceso, muchos proveedores
desarrollan un **SDK** (Software Development Kit), que suelenser una, o
varias, librerías que facilitan la integración con la API.

En este ejemplo vamos a obetener precios desde la api de Yahoo Finance,
para esto ocuparemos su librería de python (SDK) `yfinance`


```python
# importar la librería
import yfinance as yf

# lista de tickers de instrumentos financieros
tickers = ["SOXX","IVV"]

# función que descarga los precios de cierre de los tickers y los retorna en un DataFrame
df = yf.download(tickers,start="2025-01-01",end="2025-12-31")["Close"]
```

## Dataframes: Media y Mediana


```python
# Importar librería
import pandas as pd

# Cargar información desde archivo .csv
df = pd.read_csv("precios_etf.csv", sep=";", index_col="FECHA", parse_dates=True)

# Tabla pivote 
df_pivot = pd.pivot_table(data = df, index = df.index, columns = "TICKER", values = "PRECIO_CIERRE", aggfunc = "max")

# Media y mediana diaria del precio de Chile (ECH)
df_cl = df_pivot[["ECH"]]

media_cl = df_cl.mean()
mediana_cl = df_cl.median()
mediana_cl
```

## Dataframes: Varianza y Desviación Estándar


```python
# Importar librería
import pandas as pd

# Cargar información desde archivo .csv
df = pd.read_csv("precios_etf.csv", sep=";", index_col="FECHA", parse_dates=True)

# Tabla pivote 
df_pivot = pd.pivot_table(data = df, index = df.index, columns = "TICKER", values = "PRECIO_CIERRE", aggfunc = "max")

# Desviación estándar y varianza diaria del precio de Chile (ECH)
df_cl = df_pivot[["ECH"]]

desv_cl = df_cl.std()
var_cl = df_cl.var()
print(var_cl)
print(desv_cl)
```

## Dataframes: Matriz de Varianza - Covarianza


```python
# Importar librería
import pandas as pd

# Cargar información desde archivo .csv
df = pd.read_csv("precios_etf.csv", sep=";", index_col="FECHA", parse_dates=True)

# Tabla pivote 
df_pivot = pd.pivot_table(data = df, index = df.index, columns = "TICKER", values = "PRECIO_CIERRE", aggfunc = "max")

# Ordenar precios de más antiguo en adelante
df_pivot.sort_index(ascending = True, inplace = True)

# Dataframe de retornos diarios
df_ret = df_pivot.pct_change().dropna() # Calcular variación porcentual diaria y eliminar dato "NaN" de primera fila

# Matriz varianza-covarianza de retornos
df_cov = df_ret.cov()
df_cov
```

## Ejercicio 1

De tu portafolio asignado:

1)  descarga los precios histtóricos de los activos para 2024 y 2025.
2)  Crea un Dataframe con los retornos diarios.


```python
# Importar librería
import pandas as pd

# Cargar información desde archivo .csv
df = pd.read_csv("precios_etf.csv", sep=";", index_col="FECHA", parse_dates=True)

# Tabla pivote 
df_pivot = pd.pivot_table(data = df, index = df.index, columns = "TICKER", values = "PRECIO_CIERRE", aggfunc = "max")

# Filtrar desde 2024
df_filtered = df_pivot.loc["2024-01-02": "2025-12-30",:] # Filtrar el índice desde el valor 2024-01-02 hasta el final y traer todas las columnas

# Retornos diarios
df_filtered.sort_index(ascending = True, inplace = True)
df_ret = df_filtered.pct_change().dropna()
df_ret
```
