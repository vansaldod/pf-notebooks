## Tipos de datos básicos

## Operaciones numéricas

  Operador   Operación
  ---------- ---------------------------
  `+`, `-`   Suma y resta
  `*`, `/`   Multiplicación y división
  `**`       Potencia


```python
ingresos = 10000
gastos = ingresos * 0.5 # los gastos son el 50% de los ingresos.
margen = ingresos - gastos
print('El margen es: ', margen)

#  fisher
nominal = 0.05
real = 0.02
inflacion = ( (1+nominal)/ (1+real) )-1
print('La inflacion implica es: ', inflacion)
```

## Texto y resultados legibles

Una **f-string** combina texto y valores: `f"Capital: {capital}"`.

El formato **`:.2f`** muestra dos decimales.


```python
#  fisher
nominal = 0.05
real = 0.02
inflacion = ( (1+nominal)/ (1+real) )-1
reporte = f'La inflación es: {inflacion*100:.2f}%'
print(reporte)
```

## Acceder a los elementos de una lista

Las posiciones comienzan en **`0`**.

Considere una variable `precios` que es una lista con los precios
históricos de una acción.

- **`precios[0]`**: primer elemento.
- **`precios[-1]`**: último elemento.
- **`precios[0:2]`**: posiciones 0 y 1; el extremo final se excluye.


```python
precios = [10, 12, 12.5, 11, 10]
print(precios[0]) # ver el primer elemento de la lista
print(precios[-1]) # ver el último elemento de la lista
print(precios[1:4]) # los elementos entre la segunda posicioón y la cuarta
```

## Diccionarios: claves y valores

## Bucle for

Un ciclo **`for`** recorre una secuencia y repite instrucciones.


```python
precios = [1,2,3,4,5,6]
for i in precios:
  precio_qrt = i**2
  print(precio_qrt)
print('bucle realizado')
```

## Comparaciones y lógica

  Expresión            Pregunta
  -------------------- ----------------------------
  `a == b`, `a != b`   ¿Son iguales o diferentes?
  `a > b`, `a >= b`    ¿Es mayor o mayor o igual?
  `a < b`, `a <= b`    ¿Es menor o menor o igual?

**`and`**, **`or`** y **`not`** combinan condiciones.

**En clase:** comprueba si el valor de una posición supera un límite de
1 000.


```python
a = 1000
b = 2000
print(a==b) # igual
print(a!=b) # desigual
print(a>b) # mayor
print(a>=a) # mayor o igual
print(a<b) # menor
print(b<=a) # menor o igual
```

## Condiciones: if, elif y else

- **`if`** ejecuta un bloque si se cumple una condición.
- **`elif`** comprueba otra condición.
- **`else`** cubre los demás casos.


```python
duracion=5
target=6
limite=6.5

if duracion > target:
  delta = duracion - target
  if duracion > limite:
    exceso = duracion - limite
    print(f'Limite regulatorio excedido en {exceso}')
  else:
    print(f'Alerta de duracion en {delta}')
elif duracion <= target:
  print('Duracion en rango')

```
