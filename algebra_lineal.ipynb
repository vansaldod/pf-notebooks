## Repaso matrices y vectores

Una **matriz** organiza números en filas y columnas. Sus dimensiones se
escriben **$m \times n$**: **$m$ filas** y **$n$ columnas**.

Un **vector** es una colección ordenada de números. Un vector de **$n$
elementos** puede representarse como:

- **Vector fila:** una fila y $n$ columnas, **$1 \times n$**.
- **Vector columna:** $n$ filas y una columna, **$n \times 1$**.

En esta representación, los vectores son matrices con una sola fila o
una sola columna.



## Producto punto matrices y vectores

Representamos los vectores como **fila o columna**. Para multiplicar,
las **columnas del primer factor** deben coincidir con las **filas del
segundo**.



**Regla común:** $(m \times n)(n \times p) \rightarrow (m \times p)$. Si
las dimensiones interiores no coinciden, el producto no está definido.

## Trasponer matrices y vectores

**Trasponer** intercambia filas y columnas. Se aplica tanto a matrices
como a vectores representados como fila o columna, y se escribe $A^{T}$
o $u^{T}$.



- **Matriz:** sus dimensiones cambian de $m \times n$ a $n \times m$.
- **Vector:** pasa de fila ($1 \times n$) a columna ($n \times 1$), o
  viceversa.

## Introducción a NumPy

**NumPy** es una librería de Python para trabajar con datos numéricos y
realizar operaciones matemáticas.

Su estructura principal es el **array** (`ndarray`), que puede tener uno
o varios ejes. En esta clase usaremos:

- **Arrays de un eje (1D)** para representar vectores.
- **Arrays de dos ejes (2D)** para representar matrices.

La convención para importar la librería es:


```python
import numpy as np
```

Los elementos de un array comparten un tipo de dato. Usaremos arrays
numéricos.

## Producto punto entre dos arrays

Para calcular el producto punto utilizaremos la función `np.dot()`


```python
import numpy as np
vector = np.array([[1, 2, 3]])
matrix = np.array([7, 9, 11], [34,5,61])
```

```python
import numpy as np
vector = np.array([[1, 2, 3]])
matrix = np.array([[7, 9, 11], [34,5,61]])
matrix = matrix.T

# Ver dimensiones 
print("Dimensión vector", vector.shape)
print("Dimensión de la matriz", matrix.shape)

# Producto punto
result = np.dot(vector, matrix)
print(result)
print(result.shape)
```
